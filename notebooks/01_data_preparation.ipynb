The dataset description provided by Kaggle defines the purpose of each file. 
Based on this description, the grain of each dataset was subjected for verification using the actual data.

```python
# `calendar.csv` — Contains information about the dates on which the products are sold.

print("Rows:", len(calendar))
print("Unique dates:", calendar["date"].nunique())
print("Unique d values:", calendar["d"].nunique())
print("Unique calendar weeks:", calendar["wm_yr_wk"].nunique())

# Check whether one date maps to exactly one d value.
date_d_check = calendar.groupby("date")["d"].nunique()

print("\nDates mapped to more than one d value:",
      (date_d_check > 1).sum())

# Check whether one d value maps to exactly one date.
d_date_check = calendar.groupby("d")["date"].nunique()

print("d values mapped to more than one date:",
      (d_date_check > 1).sum())
```

```python
# `sales_train_validation.csv` 
# — Contains historical daily unit sales data per product and store (`d_1` to `d_1913`).

# Check whether each item-store combination appears only once.
item_store_counts_validation = (
    sales_validation
    .groupby(["item_id", "store_id"])
    .size()
)

print(
    "Duplicate item-store combinations:",
    (item_store_counts_validation > 1).sum()
)
```

```python
# `sell_prices.csv` — Contains information about the price of products sold per store and week.

print("Rows:", len(sell_prices))

# Count unique store-item-week combinations.
unique_price_combinations = (
    sell_prices[
        ["store_id", "item_id", "wm_yr_wk"]
    ]
    .drop_duplicates()
    .shape[0]
)

print(
    "Unique store-item-week combinations:",
    unique_price_combinations
)

print(
    "Duplicate store-item-week combinations:",
    len(sell_prices) - unique_price_combinations
)
```

### 5.10 Outliers / Anomalies

#### Sales demand

```python
# Identify unusually high demand values using the IQR method
for name, df in sales_datasets.items():
    demand = df.iloc[:, 6:]

    Q1 = demand.stack().quantile(0.25)
    Q3 = demand.stack().quantile(0.75)
    IQR = Q3 - Q1

    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (demand > upper_limit).sum().sum()

    print(f"{name} - Upper outlier limit:", upper_limit)
    print(f"{name} - Number of high-demand observations:", outlier_count)
```

Sales demand contains a large number of zero and low-volume observations, making a global IQR-based outlier threshold unsuitable.

```python
# Extreme Value Check
# Check the maximum daily sales value in each sales dataset to identify the highest observed demand.

for name, df in sales_datasets.items():
    demand = df.iloc[:, 6:]
    max_sales = demand.max().max()

    print(f"{name} - Maximum daily sales:", max_sales)
```

The maximum daily sales value is 763 units in both the validation and evaluation datasets. This value is retained because a high demand value is not necessarily a data-quality issue.

#### Selling Price 

```python
# Use the Interquartile Range (IQR) method to identify unusually high or low selling prices.

Q1 = sell_prices["sell_price"].quantile(0.25)
Q3 = sell_prices["sell_price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = sell_prices[
    (sell_prices["sell_price"] < lower_limit) |
    (sell_prices["sell_price"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of potential price outliers:", len(outliers))
```

The IQR method identified 303,640 selling-price observations above the upper statistical limit of $11.33. 

```python
# Minimum and Maximum Selling Price

print("Minimum selling price:", sell_prices["sell_price"].min())
print("Maximum selling price:", sell_prices["sell_price"].max())
```

Selling prices range from $0.01 to $107.32. The IQR method identified prices above $11.33 as potential statistical outliers. These observations were retained because extreme prices may represent legitimate product pricing rather than data-quality errors.

## 8. Data Transformation

Data transformation is performed only when the current data structure is not convenient for the planned analysis.

The validation and evaluation sales datasets store daily demand across separate columns (`d_1`, `d_2`, ...). This wide format is inconvenient for time-based analysis and forecasting. Therefore, the daily demand columns will be reshaped into a long format with one row per product, store, and day.

The calendar and selling-price datasets will remain as separate tables because their existing structures are suitable for analysis and can be joined when required.

#### Transform Validation Sales from Wide to Long

```python
# Reshape historical sales

id_columns = [
    "id",
    "item_id",
    "dept_id",
    "cat_id",
    "store_id",
    "state_id"
]

demand_columns = sales_validation.columns[6:]

sales_historical = sales_validation.melt(
    id_vars=id_columns,
    value_vars=demand_columns,
    var_name="d",
    value_name="demand"
)
```

##### Historical Sales Validation after Transformation

```python
print(sales_historical.shape)
print(sales_historical.columns.tolist())
sales_historical.head()
```

```python
print(
    "Duplicate product-store-day records:",
    sales_historical.duplicated(
        subset=["item_id", "store_id", "d"]
    ).sum()
)

print(
    "First day:",
    sales_historical["d"].min()
)

print(
    "Last day:",
    sales_historical["d"].max()
)

print(
    "Expected rows:",
    len(sales_validation) * len(demand_columns)
)

print(
    "Actual rows:",
    len(sales_historical)
)
```

d_999 is not the last day. This is a lexicographic string-order issue: for strings, d_999 sorts after d_1913, even though numerically 999 < 1913

```python
# Verify the actual d range numerically

d_numbers = sales_historical["d"].str.replace("d_", "").astype(int)

print("First day:", d_numbers.min())
print("Last day:", d_numbers.max())
print("Number of unique days:", d_numbers.nunique())
```

```python
print("Original product-store records:", len(sales_validation))
print(
    "Transformed product-store records:",
    sales_historical[["item_id", "store_id"]].drop_duplicates().shape[0]
)

print(
    "Original total demand:",
    sales_validation.iloc[:, 6:].sum().sum()
)

print(
    "Transformed total demand:",
    sales_historical["demand"].sum()
)

print(
    "Same item-store combinations:",
    sales_historical[["item_id", "store_id"]].drop_duplicates().shape[0]
    == sales_validation[["item_id", "store_id"]].drop_duplicates().shape[0]
)

print(
    "Same total demand:",
    sales_historical["demand"].sum()
    == sales_validation.iloc[:, 6:].sum().sum()
)
```

The wide-to-long transformation preserved the original validation sales data. The transformed dataset contains the expected 58,327,370 rows, representing 30,490 product-store combinations across 1,913 days.

The transformed data therefore preserves the original records and demand values while converting the dataset into a long format suitable for time-based demand analysis.

#### Transform Evaluation Sales from Wide to Long

The evaluation dataset contains both historical demand (`d_1` to `d_1913`) and the additional evaluation period (`d_1914` to `d_1941`).

Only the evaluation period will be reshaped into long format because the historical period has already been transformed into `sales_historical`.

```python
evaluation_demand_columns = [
    col for col in sales_evaluation.columns
    if col.startswith("d_") and 1914 <= int(col.split("_")[1]) <= 1941
]

sales_evaluation_long = sales_evaluation.melt(
    id_vars=id_columns,
    value_vars=evaluation_demand_columns,
    var_name="d",
    value_name="demand"
)
```
